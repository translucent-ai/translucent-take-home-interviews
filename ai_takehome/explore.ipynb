```python
import argparse, pandas as pd, pathlib, json, re
from sklearn.feature_extraction.text import TfidfVectorizer
from sklearn.metrics.pairwise import cosine_similarity
from sentence_transformers import SentenceTransformer

DATA_PATH = "data/denials.csv"
```

```python
df = pd.read_csv(DATA_PATH)
docs = df.apply(lambda r: f"Department {r['department']} denied because {r['denial_reason']} on {r['service_date']}", axis=1).tolist()
```

```python
pd.set_option('display.max_rows', 100)  # show all rows, no ...
counts = (
    df.groupby(['department', 'denial_reason'])
    .size()
    .reset_index(name='count')
    .sort_values(['department', 'count'], ascending=[True, False])
    .set_index(['department', 'denial_reason'])
)
counts
```
